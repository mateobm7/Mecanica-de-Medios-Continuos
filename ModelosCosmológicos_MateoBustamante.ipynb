4. **La ecuación de aceleración**. Vamos ahora e estudiar la solución

   $$
  
  Para ello integre numéricamente la ecuación de aceleración asumiendo distintos escenarios de materia.
  
  - $\Omega_0 = 0.3$, $\Omega_{0\Lambda} = 0$ (universo abierto sin energía oscura).
  - $\Omega_0 = 1$, $\Omega_{0\Lambda} = 0.7$ (universo crítico con alta energía oscura).
  
  donde $\Omega_0 = \Omega_{0m} + \Omega_{0\Lambda} = (\rho_{0m} + \rho_{0\Lambda})/\rho_{0c}$ es el parámetro de densidad total. Recuerde que por conservación de la energía $\rho_m = \rho_{0m}(a_0/a)^3$ y, por hipótesis, $\rho_\Lambda$ es constante.

  En cada escenario haga gráficos de la evolución de $a(t)$, $\ddot a(t)$, $\rho(t)$ y $H(t)$.

  Para ello se sugiere integrar la ecuación hacia atrás en el tiempo por un tiempo $\tau$ primero pequeño y después aumentarlo hasta que se llegue a $a(t)=0$.

Realice ahora las siguientes deducciones teóricas:

5. **Loopback time**. Hay una forma de calcular la edad del
   universo, que en los experimentos anteriores tuvo que encontrarse manualmente.  Para ello se define el que se conoce como el *loopback time* (tiempo de mirada atrás), $t_{LB}$. Este número es el tiempo que se ha expandido el universo desde un valor específico del factor de escala $a(t)$ hasta el presente cuando $a(t_0)=1$.

   Partiendo de la definición del parámetro de Hubble:
   $$
   H=\frac{\dot a}{a}
   $$

   **mostrar** que el tiempo cósmico correspondiente a un determinado valor de $a$ se puede obtener como:

   $$
   t_{LB}(a) = t_0 - t(a) = \int_{a}^{1} \frac{\mathrm{d}a'}{a' H(a')}
   $$
   donde $t_0$ es el tiempo actual. Nótese que, por definición, cuando $a=0$ el $t_{LB}$ es la edad $\tau_0$ del universo.

6. **Evolución de H**. Para calcular el $t_{LB}$ es necesario encontrar
   una ecuación que nos de $H$ como función de $a$. Eso se logra despejando $H$ de la ecuación de Friedman-Lemeitre:

   $$
   \epsilon=\frac{4 \pi}{3} G a^2\left[\frac{3 H^2}{8 \pi G}-\left(\rho+\rho_{\Lambda}\right)\right]
   $$
   donde recordemos $\epsilon$ es una constante.

   **Demuestre** que $H$ se puede escribir como:

   $$
   \frac{H^2}{H_0^2} = \frac{\rho+\rho_\Lambda + \rho_k}{\rho_{c0}}
   $$

   donde:
   $$
   \rho_k \equiv \frac{3 \epsilon}{4\pi G} \frac{1}{a^2}
   $$
   es una densidad ficticia que llamaremos *densidad de curvatura* (por un término usado en relatividad general, pero que aquí es solo una definición).

   Sabemos que la materia tiene una densidad que cambia según la ley $\rho_m=\rho_{0m}(a_0/a)^3$, que la energía oscura tiene $\rho_\Lambda = \rho_{0\Lambda}$ (no cambia en el tiempo) y que la densidad de curvatura varia según $\rho_k=\rho_{0k}(a_0/a)^2$. Usando esto mostrar que $H$ se puede escribir como:

   $$
   H(a) = H_0\sqrt{\Omega_{0m}(a_0/a)^3 + \Omega_{0k}(a_0/a)^2 + \Omega_\Lambda}
   $$

7. **Edad del universo teórica**. En astrofísica, el corrimiento al rojo
   se relaciona con el factor de escala así: $a/a_0 = 1/(1+z)$.  
   
   **Demuestre**, teniendo en cuenta la ecuación para $H(a)$ y escribiéndola en términos de $z$ el $t_{LB}$ es:

   $$
   t_{LB}(z) = t_H \int_0^z \frac{dz}{(1+z)\sqrt{\Omega_{0m}(1+z)^3 + \Omega_{0k}(1+z)^2 + \Omega_\Lambda}}
   $$

   De allí, la edad del universo será finalmente:

   $$
   \tau_0 = t_H \int_0^\infty \frac{dz}{(1+z)\sqrt{\Omega_{0m}(1+z)^3 + \Omega_{0k}(1+z)^2 + \Omega_\Lambda}}
   $$

   Donde $t_H=1/H_0$.

   Use la rutina `scipy.integrate.quad` para hacer esta integral (como limite superior puede poner `np.inf` y la rutina entiende que es una integral impropia) y calcular así la edad que tiene hoy el universo en los modelos cosmológicos estudiados en 3.

#**PUNTO 1**

Unidades canónicas de la cosmología.

Crearemos el siguiente sistema de ecuaciones para que las constantes principales de la cosmología sean iguales a 1.

$$
G = \frac{U_L^3}{U_M U_T^2} \ \ \ ; \ \ \ c = \frac{U_L}{U_T} \ \ \ ; \ \ \ H_0 = \frac{1}{U_T}
$$

Puedo hallar $U_T$ como:

$$
U_T = \frac{1}{H_0}
$$

Con $U_T$ conocido puedo hallar $U_L$ así:

$$
U_L = c \ U_T
$$

Como $U_L$ es conocido puedo hallar $U_M$ de la siguiente manera:

$$
U_M = \frac{c^2 U_L}{G}
$$

tenemos los valores provisionales para $G$, $c$ y $H_0$ que son los actualmente conocidos, los dejaremos como provisionales porque luego los convertiremos a $1$

```python
G_pro = cte.G #m^3/(kg s^2)

c_pro = cte.c #m/s

cte.year #Años a segundos
H0_pro = 1/(14000*1E6*cte.year) #1/s
```

Determinamos las unidades de tiempo (U_T), de longitud (U_L), de masa (U_M) y de densidad (U_rho).

```python
U_T = 1/H0_pro

U_L = c_pro*U_T

U_M = c_pro**2*U_L/G_pro

U_rho = U_M/U_L**3

print("Unidades de tiempo: ",U_T)
print("Unidades de longitud: ",U_L)
print("Unidades de masa: ",U_M)
print("Unidades de densidad: ",U_rho)
```

Con esto entonces:

```python
G = 1
c = 1
H0 = 1
```

#**PUNTO 2**

Tenemos el siguiente sistema de ecuaciones diferenciales para $H(t)$ y $\rho(t)$.

$$
  \begin{aligned}
\dot{\rho} & =-3 \rho H(t) \\
\dot{H} & =-\frac{4 \pi}{3} G \rho(t) - H^2
\end{aligned}
  $$

Planteamos el sistema de ecuaciones diferenciales.

```python
def edo_universo(Y,t):
  rho,H = Y
  drhodt = -3*rho*H
  dHdt = -4/3*np.pi*G*rho-H**2

  return [drhodt,dHdt]
```

Solución al sistema de ecuaciones diferenciales.

```python
rho_ini = 3e-27/U_rho #en unidades canónicas
H0_ini = 1 #en unidades canónicas
t=np.linspace(0,1,1000)
#condiciones iniciales
Yo=[rho_ini,H0]
ts=t
solucion_fut=odeint(edo_universo,Yo,ts)
```

```python
#Densidad en el futuro
rho_fut = solucion_fut[::,0]

#H en el futuro
H_fut = solucion_fut[::,1]
```

```python
fig, ax = plt.subplots(1,2,figsize=(9, 4.5), tight_layout=True)

ax[0].set_title("Cambio de la densidad en el futuro [uni. canónicas]")
ax[0].plot(t,rho_fut)
ax[0].set_xlabel("Tiempo")
ax[0].set_ylabel("Densidad")
ax[0].grid()


ax[1].set_title(r"Cambio de $H$ en el futuro [u. canónicas]")
ax[1].plot(t,H_fut)
ax[1].set_xlabel("Tiempo")
ax[1].set_ylabel(r"H")
ax[1].grid()

```

#**PUNTO 4**

Tenemos la ecuación de Friedmann-Lemaitre:

$$
   \ddot{a}=\frac{4 \pi}{3} G\left(2 \rho_{\Lambda}-\rho_m\right) a
   $$

por conservación de la energía $\rho_m = \rho_{0m}(a_0/a)^3$ y que $\rho_{\Lambda}$ es constante nos queda entonces que:

$$
   \ddot{a}=\frac{4 \pi}{3} G\left(2 \rho_{0\Lambda}-\rho_{0m}(a_0/a)^3\right) a
   $$

Para resolverla numericamente en python necesitamos linealizarla, esto lo haremos con los siguientes cambios de variable:

$x_1 = a$

$x_2 = \dot{a}$

Nos quedarán entonces las siguientes dos ecuaciones:
$$\dot{x_1} = x_2$$
$$
   \dot{x_2}=\frac{4 \pi}{3} G\left(2 \rho_{0\Lambda}-\rho_{0m}(a_0/x_1)^3\right) x_1
   $$


---

Creamos una función para resolver el sistema y realizar las diferentes gráficas.

```python
def sol_fri_lemaitre(rho_om,rho_olambda,t_ini):
  a0 = 1

  def fri_lemaitre(S, x1):
    x1, x2 = S

    return [x2, 4*np.pi/3*G*(2*rho_olambda - rho_om*(a0/x1)**3)*x1]

  pos_ini = (1,0)
  sol = odeint(fri_lemaitre, y0=pos_ini, t=t_ini)

  a = sol.T[0]
  a_punto = sol.T[1]

  ##-----------------------------------------------------------------
  ###Calculamos las demás variables
  #segunda derivada de a
  ddot_a = (4*np.pi/3)*G*(2*rho_olambda - rho_om*(a0 / a)**3)*a
  #rho
  rho_m = rho_om*(a0 / a)**3
  #H
  H = a_punto/a
  ##------------------------------------------------------------------

  #Graficamos
  fig, ax = plt.subplots(2,2, figsize=(9,9), tight_layout=True)
  fig.suptitle("'El tiempo estará en Unidades canónicas'")

  ax[0,0].set_title(r"Evolución de $a$ en el tiempo")
  ax[0,0].plot(t_ini,a)
  ax[0,0].set_xlabel("Tiempo")
  ax[0,0].set_ylabel(r"$a$")
  ax[0,0].grid()

  ax[0,1].set_title(r"Evolución de $\ddot{a}$ en el tiempo")
  ax[0,1].plot(t_ini,ddot_a)
  ax[0,1].set_xlabel("Tiempo")
  ax[0,1].set_ylabel(r"$\ddot{a}$")
  ax[0,1].grid()

  ax[1,0].set_title(r"Evolución de $\rho_m$ en el tiempo")
  ax[1,0].plot(t_ini,rho_m)
  ax[1,0].set_xlabel("Tiempo")
  ax[1,0].set_ylabel(r"$\rho_m$")
  ax[1,0].grid()

  ax[1,1].set_title(r"Evolución de $H$ en el tiempo")
  ax[1,1].plot(t_ini,H)
  ax[1,1].set_xlabel("Tiempo")
  ax[1,1].set_ylabel(r"$H$")
  ax[1,1].grid()
```

primer escenario $\Omega_0 = 0.3$, $\Omega_{0\Lambda} = 0$ (universo abierto sin energía oscura).
Esto quiere decir que $\rho_{0m}= 0.3\rho_{0c},$ $\rho_{0Λ} = 0$

```python
#Tiempo de integación hallado por tanteo
t_ini = np.linspace(0,-2.86,1000)


rho_oc = (3*H0**2)/(8*np.pi*G)
rho_om = 0.3*rho_oc #en unidades canónicas
rho_olambda = 0 #en unidades canónicas

sol_fri_lemaitre(rho_om,rho_olambda,t_ini)
```

Segundo escenario $\Omega_0 = 1$, $\Omega_{0\Lambda} = 0.7$, Esto quiere decir que $\rho_{0m}= 0.3\rho_{0c}, \ $ $\rho_{0Λ} = 0.7\rho_{oc}$

```python

rho_om = 0.3*rho_oc #en unidades canónicas
rho_olambda = 0.7*rho_oc #en unidades canónicas

sol_fri_lemaitre(rho_om,rho_olambda,t_ini)
```

 Además el signo menos me intercambiará los límites de integración, quedando entonces:

  $$
   t_{LB}(z) = t_H \int_0^z \frac{dz}{(1+z)\sqrt{\Omega_{0m}(1+z)^3 + \Omega_{0k}(1+z)^2 + \Omega_\Lambda}}
   $$

Para la integración se tomará $ϵ =0$ por tanto el parámetro $Ω_{0k}$ desaparecerá.

creamos una función con la integración que dependerá de $\Omega_{0m}$ y $\Omega_\Lambda$ para evaluar los diferentes modelos cosmológicos.

```python
def t_LB(Omega_om,Omega_olambda):

  tH = 1/(H0)
  t_lb = quad(lambda z: tH/((1+z)*(Omega_om*(1+z)**3 + Omega_olambda)**(1/2)), 0, np.inf)[0]

  return t_lb
```

Usaremos el primer modelo cosmológico visto $\Omega_0 = 0.3$, $\Omega_{0\Lambda} = 0$, esto quiere decir que: $\Omega_{0m} = 0.3$ $\Omega_{\Lambda} = 0$

```python
Omega_lambda = 0
Omega_om = 0.3

tiempo1_universo = t_LB(Omega_om,Omega_lambda)

#pasamos de unidades canónicas a años
tiempo1_universo_años = tiempo1_universo*U_T/cte.year

print("La edad del Universo es:",f"{tiempo1_universo_años:.2f}","años")
```

Ahora usaremos el segundo modelo cosmológico $\Omega_0 = 1$, $\Omega_{0\Lambda} = 0.7$, esto quiere decir que: $\Omega_{0m} = 0.3$ $\Omega_{\Lambda} = 0.7$

```python
Omega_lambda = 0.7
Omega_om = 0.3

tiempo2_universo = t_LB(Omega_om,Omega_lambda)

#pasamos de unidades canónicas a años
tiempo2_universo_años = tiempo2_universo*U_T/cte.year

print("La edad del Universo es:",f"{tiempo2_universo_años:.2f}","años")
```
